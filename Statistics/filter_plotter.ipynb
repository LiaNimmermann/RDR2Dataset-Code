```python
# Load the dictionary of ID to metric values

with open("id_metric_dict.pkl", "rb") as f:
    metrics = pickle.load(f)
print(metrics)
```

```python

# plot histogram of metrics
plt.hist(list(metrics.values()), bins=25)
# log y axis
plt.yscale("log")

plt.show()
```

```python
# Print IDs with metric above threshold and some example metrics for specific IDs
def check_metrics(metrics, threshold, gt=True):
    ids_to_save = [id for id, div in metrics.items() if (gt and div > threshold) or (not gt and div < threshold) ]
    if gt:
        print("IDs with metric above " + str(threshold))
    else:
        print("IDs with metric below " + str(threshold))
    print(str(len(ids_to_save)) + " out of " + str(len(metrics)) + " total IDs")
    print(str(ids_to_save))
    print("Different Scene example 1000101:", metrics[1000101])
    print("Different Scene example 1000171:", metrics[1000171])
    print("Correct example 1000166:", metrics[1000166])
    print("Correct example 1000033:", metrics[1000033])
    print("Wrong Times example 1000019:", metrics[1000019])
    print("Wrong Times example 1000812:", metrics[1000812])
    return ids_to_save
```

```python
# Save IDs to PNGs for manual inspection
path_for_inspection = r"/media/lnimmermann/T5 EVO/RDR2_dataset_processed/HDR_exr"

save_exr_images_to_png(ids_to_save, path_for_inspection)
```
