```python
import json
%pip install matplotlib
import matplotlib.pyplot as plt
```

```python
with open('gt_Fine_labelIds_mapping.json', 'r') as file:
    data = json.load(file)
    print(data)
```

```python
# Extract the keys and values into separate lists
label_ids = list(data.keys())
```

```python
print(label_ids)
```

```python
#remove first element
label_ids = label_ids[1:]
print(label_ids)
```

```python
with open('stats_summary_after.json', 'r') as file:
    stats = json.load(file)
    print(stats)
```

```python
class_counts = stats['class_counts']
print(class_counts)
```

```python
# Get counts for each label id as a new dictionary
dict_counts = {}
for label_id in label_ids:
    count = class_counts.get(label_id, 0)
    dict_counts[label_id] = count
print(dict_counts)
```

```python
#visualize as bar chart

labels = list(dict_counts.keys())
counts = list(dict_counts.values())
plt.figure(figsize=(18, 6))
plt.bar(labels, counts)
plt.xlabel('Label IDs')
plt.ylabel('Counts')
plt.title('Class Counts for Each Label ID')
plt.xticks(rotation=90)
plt.show()
```

```python
# collect all label ids with zero counts
zero_count_labels = []

for label_id, count in dict_counts.items():
    if count == 0:
        print(f'Label ID with zero count: {label_id}')
        zero_count_labels.append(label_id)
```

```python
print(zero_count_labels)
```

```python
# collect all label ids with low counts
low_count_labels = []

for label_id, count in dict_counts.items():
    if count < 50:
        print(f'Label ID: {label_id} Count: {count}')
        low_count_labels.append(label_id)
```

```python
with open('animal_wordnet_mapping.json', 'r') as file:
    animal_mapping = json.load(file)
    print(animal_mapping)
```

```python
# create lists of superclasses, subclasses and synset_names
superclasses = []
subclasses = []
synset_names = []
offset_ids = []
for label_id in label_ids:
    if label_id in animal_mapping:
        mapping = animal_mapping[label_id]
        superclasses.append(mapping[0])
        subclasses.append(mapping[1])
        synset_names.append(mapping[2])
        offset_ids.append(mapping[3])
        
        
# unique values in superclasses and subclasses
superclasses = list(set(superclasses))
subclasses = list(set(subclasses))


print(superclasses)
print(subclasses)
print(synset_names)
print(offset_ids)
```

```python
# size of superclasses and subclasses
print(superclasses.__len__())
print(subclasses.__len__())
```

```python
# superclasses and subclasses count
superclass_counts = {}
subclasses_counts = {}
synset_counts = {}
for superclass in superclasses:
    superclass_counts[superclass] = 0
    
for subclass in subclasses:
    subclasses_counts[subclass] = 0
    
for synset_name in synset_names:
    synset_counts[synset_name] = 0
    
for label_id in label_ids:
    if label_id in animal_mapping:
        mapping = animal_mapping[label_id]
        superclass = mapping[0]
        subclass = mapping[1]
        synset_name = mapping[2]
        count = dict_counts.get(label_id, 0)
        superclass_counts[superclass] += count
        subclasses_counts[subclass] += count
        synset_counts[synset_name] += count
        
print(superclass_counts)
print(subclasses_counts)
print(synset_counts)
```

```python
# visualize superclass counts as bar chart
labels = list(superclass_counts.keys())
counts = list(superclass_counts.values())
plt.figure(figsize=(12, 6))
plt.bar(labels, counts)
plt.xlabel('Superclasses')
plt.ylabel('Counts')
plt.title('Superclass Counts')
plt.show()
```

```python
# visualize subclass counts as bar chart
labels = list(subclasses_counts.keys())
counts = list(subclasses_counts.values())
plt.figure(figsize=(12, 6))
plt.bar(labels, counts)
plt.xlabel('Subclasses')
plt.ylabel('Counts')
plt.xticks(rotation=90)
plt.title('Subclass Counts')
plt.show()
```

```python
# visualize synset counts as bar chart
labels = list(synset_counts.keys())
counts = list(synset_counts.values())
plt.figure(figsize=(20, 6))
plt.bar(labels, counts)
plt.xlabel('Synset Names')
plt.ylabel('Counts')
plt.xticks(rotation=90)
plt.title('Synset Counts')
plt.show()
```
